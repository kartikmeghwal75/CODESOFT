```python
import pandas as pd

def load_train_data(train_data):
    data = []

    with open(train_data, 'r', encoding='utf-8') as f:
        for line in f:
            parts = line.strip().split(" ::: ")

            if len(parts) >= 4:
                movie_id = parts[0]
                title = parts[1]
                genre = parts[2]
                description = " ::: ".join(parts[3:])

                data.append([movie_id, title, genre, description])

    return pd.DataFrame(
        data,
        columns=["id", "title", "genre", "description"]
    )

train_df = load_train_data("Genre Classification Dataset/train_data.txt")

print(train_df.head())
print(train_df.shape)
```

```python
def load_test_data(test_data):
    data = []

    with open(test_data, 'r', encoding='utf-8') as f:
        for line in f:
            parts = line.strip().split(" ::: ")

            if len(parts) >= 3:
                movie_id = parts[0]
                title = parts[1]
                description = " ::: ".join(parts[2:])

                data.append([movie_id, title, description])

    return pd.DataFrame(
        data,
        columns=["id", "title", "description"]
    )

test_df = load_test_data("Genre Classification Dataset/test_data.txt")

print(test_df.head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    stop_words='english',
    max_features=50000,
    ngram_range=(1,2)
)

X_train = tfidf.fit_transform(
    train_df['description']
)

X_test = tfidf.transform(
    test_df['description']
)

y_train = train_df['genre']
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=1000,
    n_jobs=-1
)

model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)

test_df["Predicted_Genre"] = predictions

print(
    test_df[
        ["title", "Predicted_Genre"]
    ].head()
)
```

```python
test_df["Predicted_Genre"] = predictions

test_df.to_csv(
    "movie_predictions.csv",
    index=False
)

print("Prediction file saved.")
```
